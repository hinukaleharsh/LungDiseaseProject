```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/lung_cancer_clean.csv')

df['Level_num'] = df['Level'].map({'Low': 0, 'Medium': 1, 'High': 2})

corr = (df.drop(columns='Level').corr()['Level_num']
          .drop('Level_num').sort_values(ascending=False))
print(corr.round(2))

corr.plot(kind='barh', figsize=(8, 8))
plt.title('Correlation of each feature with risk level')
plt.tight_layout()
plt.savefig('../reports/figures/correlation_with_level.png', dpi=150)
plt.show()

print(df.groupby('Level')[['Age', 'Smoking', 'Coughing of Blood', 'Wheezing']].mean().round(2))
```
