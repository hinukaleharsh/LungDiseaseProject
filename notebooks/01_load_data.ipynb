```python
import pandas as pd

df = pd.read_csv('../data/raw/lung_cancer.csv')
df.head()
```

```python
print(df.shape)
print(df.columns.tolist())
df.info()
print(df.isnull().sum().sum())
```

```python
data = df.drop(columns=['index', 'Patient Id'])
print(data['Level'].value_counts())
print(data.drop(columns='Level').describe().T[['min', 'max']])
print('Duplicate rows:', data.duplicated().sum())
print(sorted(data['Gender'].unique()))
```

```python
data_clean = data.drop_duplicates().reset_index(drop=True)

print(data_clean.shape)
print(data_clean['Level'].value_counts())

features_only = data_clean.drop(columns='Level')
print('Same symptoms, different Level:', features_only.duplicated().sum())

data_clean.to_csv('../data/processed/lung_cancer_clean.csv', index=False)
```
